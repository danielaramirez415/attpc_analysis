# Examine bit flip
Pre-$\verb|Spyral|$ notebook that identifies bitflips across individual $\verb|COBO|\text{s}$. The bitflip is carried out across the experiment and consequently, it is ideal to implement a bitflip correction to the affected $\verb|COBO|\text{s}$. As a general comment the bitflips can occur up to $N$ number to  $N=1,...,16$ ($2^N$). 

It is a _visual_ notebook; therefore, you must ran the executable called `bitflip_hardware_id.py` first to produce a csv file with the affected cobos. It helps identify single traces with a bitflip.
